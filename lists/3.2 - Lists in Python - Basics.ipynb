```python
cashflows = [100, 120, 140, 160]
print("First cashflow:", cashflows[0]) 
print("Second cashflow:", cashflows[1])
print("Third cashflow:", cashflows[2])
print("Last cashflow:", cashflows[-1]) 
```

```python
cashflows[0] = 105 
cashflows[-1] = 200 
cashflows[2] = 150
print(cashflows)
```

```python
cashflows[0] *= 2
print(cashflows)
```

```python
cashflows = [100, 120, 140, 160]
print("All cashflows:",cashflows[0:4])
print("First three cashflows:",cashflows[0:3])
print("All cashflows except first:",cashflows[1:])
print("Last two cashflows:",cashflows[-2:])
```

```python
print("Every alternate cashflow:", cashflows[::2])
print("Reverse list:", cashflows[::-1])
print("Length of cashflows list:", len(cashflows))
```
