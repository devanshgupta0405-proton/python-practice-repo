```python
returns = [0.02, "", -0.01, 0.015, None, 0.03, "0.005"]

inv_value = 100
for r in returns:
    if type(r) != float:
        continue
    inv_value = inv_value * (1+r)
print("Total Investment Value:", inv_value)
```
