```python
buying_prices = [2450, 2475, 2490, 2500]
quantities = [10, 50, 25, 15]
selling_price = 2600

total_profit = 0 

for buy_price in buying_prices:
    profit = selling_price - buy_price
    total_profit += profit

#Calc Total Port. Profit (involving qty)

total_sell_value = selling_price * sum(quantities)
total_buy_value = 0

if len(buying_prices) == len(quantities):
    for i in range(len(buying_prices)):
        buy = buying_prices[i]
        qty = quantities[i]
        buy_value = buy * qty  
        total_buy_value += buy_value

else:
    print("Some error in Prices & Qty")

final_profit = total_sell_value - total_buy_value
print(final_profit)
```
