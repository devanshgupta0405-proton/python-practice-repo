```python
initial_investment = [100000, 200000]
growth_rates = [ 0.07, 0.08, 0.09]
investment_years = [5, 10, 15, 20]

for amount in initial_investment:
    for rate in growth_rates:
        for years in investment_years:
            acc_fund = amount * (1 + rate) ** years
            if acc_fund >= 2 * amount:
                print("Initial Investment:", amount,
                    "Yearly Return:", rate,
                    "Investment Years:", years,
                    "Final Fund Value:", acc_fund)
            else:
                print("Not enough to double the investment")
```

```python
rates = [0.05, 0.06, 0.07, 0.08]
inflation = [0.02, 0.03, 0.04]
growth = [0.03, 0.04, 0.05]

for r in rates:
    for inf in inflation:
        for g in growth:
            real_return = r - inf
            if real_return > 0.02: 
                print("Rate:", r, "Inflation:", inf, "Growth:", g, "Real Return:", real_return)
```
