```python
daily_returns = [0.025, -0.01, 0.015, 0.05, 0.02, 0.03, -0.02, -0.01, 0.015, 0.05, 0.025, 0.0375]
target_cumulative_return = 0.15

cumulative_return = 0
count_index = 0

while cumulative_return < target_cumulative_return:
    print(count_index, "respective cumulative return", cumulative_return)
    cumulative_return += daily_returns[count_index]
    count_index += 1 
```
