```python
portfolio = [
    {"asset": "Stock A", "volatility": 0.15, "P&L": 500},
    {"asset": "Stock B", "volatility": 0.08, "P&L": 2200},
    {"asset": "Stock C", "volatility": 0.35, "P&L": -1200},
    {"asset": "Stock D", "volatility": 0.12, "P&L": -1600},    
]

vol_threshold = 0.30
loss_threshold = -1500

for item in portfolio:
    checks = [
        (item["volatility"] > vol_threshold,
         f"ALERT: {item['asset']} volatility too high ({item['volatility']})"),
        (item["P&L"] < loss_threshold,
         f"ALERT: {item['asset']} loss exceeds threshold ({item['P&L']})"),
    ]

    for triggered, message in checks:
        if triggered:
            print(message)
            break
    else:
        print(f"{item['asset']} - Status: OK")
```

```python
import csv

values = []

with open(r"C:\Users\ACER\OneDrive - Nimeya Technologies Private Limited\Python Practice.csv", newline="", encoding="utf-8-sig") as file:
    for row in csv.reader(file):
        if row and row[0].strip():  # skip empty cells
            values.append(int(row[0]))  # use int(...) if all values are whole numbers

print(len(values))  # should print 252
print(values[:5])   # preview the first five
```

```python
# Check each 30-day rolling average. Stop at the first day it drops below 50,000.
thirty_day_sum = 0

for i, value in enumerate(values):
    thirty_day_sum += value

    if i >= 30:
        thirty_day_sum -= values[i - 30]

    if i >= 29:
        thirty_day_avg = thirty_day_sum / 30
        if thirty_day_avg < 50000:
            print(
                f"Day {i + 1}: 30-day average = {thirty_day_avg:.2f} - "
                "Low Balance Alert. Charge FEES!"
            )
            break
```
